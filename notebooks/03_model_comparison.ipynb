# Chronological Model Evaluation

This notebook documents and inspects reports generated by the model-training pipeline. It does not refit production models or tune decisions on the 2025 test season.

## Evaluation protocol

- **Train:** 2016–2023
- **Validation:** 2024 (select model families, weights, and feature experiments)
- **Test:** 2025 (held out for final reporting)
- **WR receiving-yards metrics:** MAE, RMSE, median absolute error, and bias.
- **Other position outcomes:** MAE, RMSE, and median absolute error per stat.
- **Rolling validation:** fit blend weights on earlier 2024 weeks, score later weeks, compare with recent-form baselines on the same player-game cohort.

All splits are chronological. Preprocessing is fitted on training rows only. Keep the 2025 holdout for final, once-only diagnostics rather than repeated tuning.

## Reproduce the reports

From the repository root, with local feature tables and model artifacts available:

```bash
uv run the-script compare-wr-models
uv run the-script evaluate-yards-challengers
uv run the-script evaluate-wr-rolling-validation
uv run the-script backtest-recency-features
uv run the-script backtest-nextgen-features
```

The first two commands regenerate model comparison/ensemble artifacts. Back up `models/` if you want to retain the currently committed models before running them.
